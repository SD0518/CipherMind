```python
from logparser.Drain import LogParser

input_dir = '/content/logparser/logparser/Drain/Test_log' # The input directory of log file
output_dir = '/content/logparser/logparser/Drain/result_log'  # The output directory of parsing results
log_file = 'Thunderbird_2k.log'  # The input log file name
log_format = '<Label> <TimeStamp> <Date> <User> <Month> <Day> <Time> <Location> <Component>: <Content>' # Define log format to split message fields
# Regular expression list for optional preprocessing (default: [])
regex = [
    r'(/|)([0-9]+\.){3}[0-9]+(:[0-9]+|)(:|)',  # IPs
    r'\[\d+\]',                                # [2915] -> <>
    r'\d{2}:\d{2}:\d{2}',                      # times
]

st = 0.5  # Similarity threshold
depth = 4  # Depth of all leaf nodes

parser = LogParser(log_format, indir=input_dir, outdir=output_dir,  depth=depth, st=st, rex=regex)
parser.parse(log_file)
import pandas as pd

# Load the structured log CSV
df = pd.read_csv("/content/logparser/logparser/Drain/result_log/Thunderbird_2k.log_structured.csv")

# Extract PID (digits inside square brackets)
df["PID"] = df["Component"].str.extract(r"\[(\d+)\]", expand=False)

# Clean the Component: drop "[digits]" and any trailing colon, then trim spaces
df["Component"] = (
    df["Component"]
      .str.replace(r"\[\d+\]", "", regex=True)   # remove [2915]
      .str.replace(r":$", "", regex=True)        # remove final colon
      .str.strip()
)

# If you want PID as numbers instead of strings, convert:
df["PID"] = df["PID"].astype("Int64")  # allows NA values

# Save to new CSV
df.to_csv("/content/logparser/logparser/Drain/result_log/Thunderbird_with_PID.csv", index=False)
```
